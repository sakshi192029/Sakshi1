```python
import math

def golden(f, a, b, tol=1e-2):
    phi = (1 + 5**0.5) / 2  # golden ratio
    while b - a > tol:
        c = b - (b - a) / phi
        d = a + (b - a) / phi
        if f(c) < f(d):
            b = d
        else:
            a = c
    return (a + b) / 2

# test function
f = lambda x: (x - 3)**2 + 2
print(golden(f, 0, 6))
```
