```python
# Interval Halving Method

import math

# Define objective function
f = lambda x: (x-2.5)**2 + 3

def interval_halving(f, a, b, tol=1e-3):
    k = 0
    while (b - a) > tol:
        k += 1
        mid = (a + b) / 2
        l = (a + mid) / 2
        r = (mid + b) / 2

        f_l, f_mid, f_r = f(l), f(mid), f(r)

        if f_l < f_mid:
            b = mid
        elif f_r < f_mid:
            a = mid
        else:
            a, b = l, r

        print(f"Iter {k}: Interval=({a:.4f}, {b:.4f})")

    return (a + b) / 2

x_opt = interval_halving(f, 0, 6)
print("Interval Halving Optimum:", x_opt, "f(x) =", f(x_opt))
```

```python
def fibF(n):
  fib = [1,1]
  for i in range(2,n):
    fib.append(fib[-1] + fib[-2])
  return fib

def fib_sear(f,a,b,n=10):
  fib = fibF(n+1)
  k = 0
  l = b-a

  # Initial test points
  x1 = a + (fib[n-2]/fib[n]) * l
  x2 = a + (fib[n-1]/fib[n]) * l
  f1, f2 = f(x1), f(x2)

  for i in range(1, n-1):
    k += 1
    if f1> f2:
      a = x1
      x1 = x2
      f1 = f2
      x2 = a + (fib[n-k-1]/fib[n-k]) * (b - a)
      f2 = f(x2)
    else :
      b = x2
      x2 = x1
      f2 = f1
      x1 = a + (fib[n-k-2]/fib[n-k]) * (b - a)
      f1 = f(x1)
    print(f"Iter {k} : Interval=({a:.4f}, {b:.4f})")
  return (x1+x2) / 2

x_opt = fib_sear(f, 0, 6, n=12)
print("Fibonacci Search Optimum:", x_opt, "f(x) =", f(x_opt))
```
