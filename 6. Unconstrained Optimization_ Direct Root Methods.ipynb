```python
# Function and Derivatives
f = lambda x: x**4 - 3*x**3 + 2
f1 = lambda x: 4*x**3 - 9*x**2
f2 = lambda x: 12*x**2 - 18*x
```

```python
def newton_root(f1, f2, x0, tol=1e-6, max_iter = 20):
  x = x0
  for k in range(max_iter):
    x_new = x - f1(x)/f2(x)
    print(f"Iter {k+1}: x={x_new: .6f}, f(x)={f(x_new): .6f}")
    if abs(x_new - x) < tol:
      return x_new
    x = x_new
  return x

x_opt = newton_root(f1, f2, 3.0)
print("\nNewton-Raphson optimun:", round(x_opt, 5), "f(X)=", round(f(x_opt), 5))
```

```python
def secant_root(f1, x0, x1, tol=1e-6, max_iter=20):
    for k in range(max_iter):
        f0, f1_val = f1(x0), f1(x1)
        if abs(f1_val - f0) < 1e-12:
            break
        x2 = x1 - f1_val*(x1-x0)/(f1_val-f0)
        print(f"Iter {k+1}: x={x2: .6f}, f(x)={f(x2): .6f}")
        if abs(x2 - x1) < tol:
            return x2
        x0, x1 = x1, x2
    return x1

x_opt = secant_root(f1, 2.0, 3.0)
print("Secant Method Optimum:", round(x_opt, 5), "f(X)=", round(f(x_opt), 5))

# Diffenret Output
# Photo
```
