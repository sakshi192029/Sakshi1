Lag

Minimize f(x) = x^2 + y^2 subject to x + 2xy = 6

```python
import sympy as sp

# Variables
x , y, lam = sp.symbols('x y lam')

# Objective Function
f = x**2 + y**2

# COnstraint
h = x + 2*y - 6

# Lagrabgiaan
L = f + lam * h

# First Order Conditon
eps = [sp.diff(L,v) for v in [x,y]] + [h]

# Solve system of equations
sol = sp.solve(eps, (x, y, lam), dict = True)
print("Solution:", sol)

# Extract first solutin dictonary
sol_dict = sol[0]

# Evalaute f(x,y) at solutin
opt_val = f.subs(sol_dict)
print("Optimal Value f(x ,y) :" , opt_val)
```
