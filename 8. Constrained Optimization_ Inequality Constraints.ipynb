```python
import numpy as np
from scipy.optimize import minimize

# Define the objective function to minimize
def objective(x):
    return x[0]**2 + x[1]**2  # Minimize x^2 + y^2

# Define the inequality constraint: x + y >= 1
def constraint_ineq(x):
    return x[0] + x[1] - 1  # For g(x) >= 0, return g(x)

# Define the equality constraint: x - y = 0
def constraint_eq(x):
    return x[0] - x[1]  # For h(x) = 0, return h(x)

# Initial guess for the variables
x0 = np.array([0.5, 0.5])

# Define bounds for the variables (optional)
bnds = ((0, None), (0, None))  # x >= 0, y >= 0

# Define the constraints
constraints = [
    {'type': 'ineq', 'fun': constraint_ineq},
    {'type': 'eq', 'fun': constraint_eq}
]

# Solve the optimization problem
solution = minimize(objective, x0, method='SLSQP', bounds=bnds, constraints=constraints)

# Print the results
print("Optimal solution (x, y):", solution.x)
print("Optimal objective value:", solution.fun)
print("Success:", solution.success)
print("Message:", solution.message)
```
